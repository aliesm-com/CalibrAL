## 6) Show final artifact paths

```python
from pathlib import Path

candidates = [
    Path("outputs"),
    Path("/content/outputs"),
]
files = [
    "results.csv",
    "predictions.csv",
    "run_summary.json",
    "comparison_summary.json",
    "learning_curve.png",
    "method_comparison.png",
    "report.md",
]

for root in candidates:
    print(f"\n== {root} exists={root.exists()} ==")
    if not root.exists():
        continue
    for name in files:
        path = root / name
        print(f"{path} | exists={path.exists()} | size={path.stat().st_size if path.exists() else 0}")
```
